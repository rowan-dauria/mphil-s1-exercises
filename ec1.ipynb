```python
def plot_scatter(x, y, title="Scatter Plot", xlabel="X-axis", ylabel="Y-axis"):
    """
    Creates a scatter plot from x and y arrays using matplotlib.

    Parameters:
        x (array-like): Data for the x-axis
        y (array-like): Data for the y-axis
        title (str): Plot title (default: "Scatter Plot")
        xlabel (str): Label for the x-axis (default: "X-axis")
        ylabel (str): Label for the y-axis (default: "Y-axis")
    """
    plt.figure(figsize=(6, 4))
    plt.scatter(x, y, alpha=0.7)
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.grid(True, linestyle="--", alpha=0.5)
    plt.tight_layout()
    plt.show()
```

```python
def montyHall(N = 3, stick = False):
    if N < 3:
        print('No')

    # 1 = win, 0 = lose
    win_lose_tracker= []

    for i in range(1000):
        your_door_choice = random.randint(0, N-1)
        winning_choice = random.randint(0, N-1)
        if (winning_choice == your_door_choice):
            if (stick == True): win_lose_tracker.append(1)
            else: win_lose_tracker.append(0)
        else:
            if (stick == True): win_lose_tracker.append(0)
            else: win_lose_tracker.append(1)


    return win_lose_tracker

montyHall(50)

print(np.array(montyHall(50)).mean())
print(np.array(montyHall(3)).mean())



```
